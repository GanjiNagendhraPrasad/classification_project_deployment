```python
import numpy as np
import pandas as pd
import sklearn
```

```python
df=pd.read_csv('/content/gender_classification_v7.csv')
df
```

```python
df['gender']=df['gender'].map({'Male':0,'Female':1}).astype(int)
```

```python
df
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
odd=np.arange(3,51,2)
odd
```

```python
a=[]
for i in odd:
  p=KNeighborsClassifier(n_neighbors=i)
  p.fit(x_train,y_train)
  ans=p.predict(x_test)
  a.append(accuracy_score(y_test,ans))

a
```

```python
odd[np.argmax(a)]
```

```python
reg=KNeighborsClassifier(n_neighbors=9)
reg.fit(x_train,y_train)
```

#train

```python
ans_train=reg.predict(x_train)
```

```python
confusion_matrix(y_train,ans_train)
```

```python
accuracy_score(y_train,ans_train)
```

```python
print(classification_report(y_train,ans_train))
```

#test

```python
ans_test=reg.predict(x_test)
```

```python
confusion_matrix(y_test,ans_test)
```

```python
accuracy_score(y_test,ans_test)
```

```python
print(classification_report(y_test,ans_test))
```
